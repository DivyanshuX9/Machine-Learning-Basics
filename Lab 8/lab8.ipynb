```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
```

```python
df=pd.read_csv("train-telecom-churn.csv")
```

```python
features=['Monthly Charge','Tenure in Months', 'Total Charges','Total Long Distance Charges']
```

```python
X=df[features]
```

```python
X_train,X_test=train_test_split(X,test_size=0.3,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
```

```python
inertia=[]
silhouette=[]
k_range=range(2,11)
```

```python
from sklearn.metrics import silhouette_score
for k in k_range:
    model=KMeans(n_clusters=k,random_state=42,n_init=10)
    labels=model.fit_predict(X_train_scaled)
    
    inertia.append(model.inertia_)
    silhouette.append(silhouette_score(X_train_scaled,labels))
```

```python
import sklearn.metrics import silhouette_score
for k in range k_range:
    model=KMeans(n_clusters=k,random_state=42,n_init=10)
    label=model.fit_predict(X_train_scaled)
    
    inertia.append(model.inertia_)
    silhouette.append(silhouette_score(X_train_scaled,label))
```

```python
plt.plot(k_range,inertia,marker='o')
plt.title("Elbow Method")
plt.xlabel("K")
plt.ylabel("Inertia")
plt.show
```

```python
plt.plot(k_range,silhouette ,marker='o')
plt.title("Silhouette Scores")
plt.xlabel('K')
plt.ylabel('Silhouette Score')
plt.show()
```

```python
best_k=k_range[np.argmax(silhouette)]
print(best_k)
```
